```python
import tensorflow as tf

# 1. Define your parameters
batch_size = 32
img_height = 224 # Standard size for most CNNs (like ResNet or VGG)
img_width = 224
dataset_dir = "../Datasets/Training"

# 2. Load the dataset directly from the directory structure
train_dataset = tf.keras.utils.image_dataset_from_directory(
  dataset_dir,
  validation_split=0.2,      # Reserve 20% of the data for validation
  subset="training",
  seed=123,                  # Ensures reproducibility
  image_size=(img_height, img_width),
  batch_size=batch_size
)

val_dataset = tf.keras.utils.image_dataset_from_directory(
  dataset_dir,
  validation_split=0.2,
  subset="validation",
  seed=123,
  image_size=(img_height, img_width),
  batch_size=batch_size
)

# 3. Optimize for performance (caches data in memory to train faster)
AUTOTUNE = tf.data.AUTOTUNE
train_dataset = train_dataset.cache().shuffle(1000).prefetch(buffer_size=AUTOTUNE)
val_dataset = val_dataset.cache().prefetch(buffer_size=AUTOTUNE)
```
